## 1. Problem definition

Stay cables carry the deck loads of a cable-stayed bridge to the pylons, so their tension is one of the most important quantities to monitor. If the tension is too high, the cable may approach its capacity; if it changes over time, it may indicate damage, loss of anchorage or changes in the structure.

A common and inexpensive way to estimate the tension of a cable in service is the **vibration method**:

1. An accelerometer is attached to the cable and records its ambient vibration (in this example, records of 200 s sampled at 50 Hz).
2. The Fast Fourier Transform (FFT) of the record is computed, and the fundamental frequency $f_1$ is obtained by averaging the five highest peaks.
3. The tension is computed with the taut-string formula, which relates the tension to the frequency, the mass per unit length and the length of the cable.

The problem is to estimate the **mean and variance of the permanent tension** of a stay cable of a cable-stayed bridge in Alberta, considering that the mass per unit length, the effective length and the measured frequency are uncertain.

## 2. Analytical equation

The output variable is the permanent tension $T$ of the cable, given by the taut-string formula:

$$T = 4\,m\,L^2\left(\frac{f_n}{n}\right)^2 \tag{1}$$

where $m$ is the total mass per unit length (kg/m), $L$ is the effective length (m), $f_n$ is the measured natural frequency of mode $n$ (Hz), and $T$ is obtained in N. In this problem the fundamental mode is used ($n = 1$), so:

$$T = 4\,m\,L^2 f_1^2 \tag{2}$$

## 3. Input variables and assumptions

The cable is made of 37 strands of 15.7 mm, grade 1860 MPa (ASTM A416-16).

| Variable | Mean | CV | Std. deviation | Source of uncertainty |
|---|---|---|---|---|
| $m$, total mass per unit length | 54 kg/m | 0.02 | 1.08 kg/m | Manufacturing tolerances of strands, sheathing and filler |
| $L$, effective length | 150 m | 0.03 | 4.50 m | Effective vibrating length affected by dampers and anchorages |
| $f_1$, fundamental frequency | 0.96 Hz | 0.01 | 0.0096 Hz | Frequency resolution, measurement noise and temperature |

**Assumptions:**

- The input variables are statistically independent, because they come from different sources (manufacturing, geometry and measurement). Therefore, the variance of the output has no covariance terms.
- The mode number $n = 1$ is identified correctly and is deterministic.
- The cable behaves as an ideal taut string: the bending stiffness, the sag (catenary effect) and the effect of the dampers are neglected.
- The mean values and coefficients of variation are illustrative values, assumed based on typical manufacturing tolerances and measurement errors.

## 4. First-order approximation

With the first-order approximation, the function $Y = g(X_1, X_2, \ldots, X_k)$ is expanded in a Taylor series around the mean values, keeping only the linear terms. For independent variables:

$$\mu_Y \approx g(\mu_{X_1}, \mu_{X_2}, \ldots, \mu_{X_k}) \tag{3}$$

$$\sigma_Y^2 \approx \sum_{i=1}^{k} \left(\frac{\partial g}{\partial x_i}\right)^2_{\mu} \sigma_{X_i}^2 \tag{4}$$

where the partial derivatives are evaluated at the mean values.

### 4.1 Mean of the tension

Applying Eq. (3) to Eq. (2):

$$\mu_T \approx 4\,\mu_m\,\mu_L^2\,\mu_f^2 = 4\,(54)\,(150)^2\,(0.96)^2 = 4\,478\,976\ \text{N} = 4479.0\ \text{kN} \tag{5}$$

### 4.2 Partial derivatives

Evaluated at the mean values:

$$\frac{\partial T}{\partial m} = 4\,L^2 f^2 = 4\,(150)^2\,(0.96)^2 = 82\,944\ \text{N per kg/m} = 82.94\ \text{kN per kg/m} \tag{6}$$

$$\frac{\partial T}{\partial L} = 8\,m\,L\,f^2 = 8\,(54)\,(150)\,(0.96)^2 = 59\,720\ \text{N/m} = 59.72\ \text{kN/m} \tag{7}$$

$$\frac{\partial T}{\partial f} = 8\,m\,L^2 f = 8\,(54)\,(150)^2\,(0.96) = 9\,331\,200\ \text{N/Hz} = 9331.2\ \text{kN/Hz} \tag{8}$$

### 4.3 Variance of the tension

Applying Eq. (4):

$$\sigma_T^2 \approx \left(\frac{\partial T}{\partial m}\right)^2 \sigma_m^2 + \left(\frac{\partial T}{\partial L}\right)^2 \sigma_L^2 + \left(\frac{\partial T}{\partial f}\right)^2 \sigma_f^2 \tag{9}$$

$$\sigma_T^2 \approx (82.94 \times 1.08)^2 + (59.72 \times 4.50)^2 + (9331.2 \times 0.0096)^2 = 89.58^2 + 268.74^2 + 89.58^2 = 88\,269\ \text{kN}^2 \tag{10}$$

$$\sigma_T \approx 297.1\ \text{kN}, \qquad CV_T = \frac{297.1}{4479.0} = 0.066 \tag{11}$$

**Check:** since Eq. (2) is a product of powers of the input variables, the first-order approximation of the coefficient of variation can also be written as:

$$CV_T^2 \approx CV_m^2 + (2\,CV_L)^2 + (2\,CV_f)^2 = 0.02^2 + (2 \times 0.03)^2 + (2 \times 0.01)^2 = 0.0044 \tag{12}$$

$$CV_T \approx 0.066 \tag{13}$$

which gives the same result as Eq. (11).

## 5. Results and conclusions

| Output | Value |
|---|---|
| Mean of the tension, $\mu_T$ | 4479.0 kN |
| Variance of the tension, $\sigma_T^2$ | 88 269 kN² |
| Standard deviation, $\sigma_T$ | 297.1 kN |
| Coefficient of variation, $CV_T$ | 0.066 |

- The tension estimated with the vibration method has a coefficient of variation of about 6.6 %, more than twice the largest coefficient of variation of the input variables. This happens because $L$ and $f_1$ appear squared in the equation, so their coefficients of variation are doubled.
- The effective length $L$ dominates the uncertainty of the tension (81.8 % of the variance), because it has the largest coefficient of variation and appears squared. Reducing the uncertainty of $L$ (for example, by defining more accurately the effective vibrating length between dampers) would be the most effective way to improve the estimate of the tension.
- The mass and the frequency contribute the same amount (9.1 % each), because $CV_m = 2\,CV_f$ and the frequency appears squared.
- The first-order approximation linearizes a nonlinear function, so the mean and variance are approximate. Since the coefficients of variation are small, the approximation is expected to be accurate.
